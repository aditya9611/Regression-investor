```python
# Install dependencies if needed:
# !pip install yfinance pandas matplotlib torch gluonts --upgrade

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from gluonts.dataset.common import ListDataset
from gluonts.torch.model.deepar import DeepAREstimator
from gluonts.torch.distributions import StudentTOutput

# Load tickers from Excel
tickers_df = pd.read_excel("stocks.xlsx")  # Column name must be 'Ticker'
tickers = tickers_df['Ticker'].tolist()

def analyze_stock(ticker):
    # Download 6 months of daily stock data
    data = yf.download(ticker, period="6mo", interval="1d", auto_adjust=True)
    if data.empty:
        print(f"No data found for {ticker}")
        return

    dates = data.index
    close_prices = data['Close'].to_numpy().ravel()  # Flatten to 1D
    current_price = close_prices[-1].item()

    # Prepare GluonTS dataset
    training_data = ListDataset(
        [{"start": dates[0], "target": close_prices}],
        freq="1D"
    )

    # DeepAREstimator (PyTorch)
    estimator = DeepAREstimator(
        prediction_length=1,
        freq="1D",
        distr_output=StudentTOutput(),
        trainer_kwargs={
            "max_epochs": 20,
            "accelerator": "cpu",  # or "gpu" if available
            "devices": 1
        }
    )

    # Train model
    predictor = estimator.train(training_data=training_data)

    # Forecast next day
    forecast = next(predictor.predict(training_data))
    predicted_price = forecast.mean[-1].item()

    # Output
    print(f"\n=== {ticker} ===")
    print(f"Current Price: {current_price:.2f}")
    print(f"Predicted Price (next day): {predicted_price:.2f}")

    # Buy recommendation + plot
    if predicted_price > current_price:
        print(f"Recommendation: Consider buying {ticker}")
        plt.figure(figsize=(10,5))
        plt.plot(dates, close_prices, label='Actual Price')
        plt.axhline(predicted_price, color='red', linestyle='--', label='Predicted Next Day Price')
        plt.title(f"{ticker} Price vs Predicted Next Day Price")
        plt.xlabel("Date")
        plt.ylabel("Price")
        plt.legend()
        plt.show()
    else:
        print(f"Recommendation: No buy signal for {ticker}")

# Run analysis for all tickers
for ticker in tickers:
    analyze_stock(ticker)
```
